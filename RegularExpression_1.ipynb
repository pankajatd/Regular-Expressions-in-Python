**Meta Characters in Regular Expressions: .^$*+?{}[]\|()**
. -Any Character Except a Newline character

^ -Starts with "^hello"

$-ends with "world$"

*-Zero or more occurances "aix*"

+ -One or more occurances "aix+"

?

{}-Exactly specified number of occurances "al{2}"

[]-A set of Characters "[a-m]"

\-Special Sequence(or escape special characters)"\d"

|-  Either or falls|stays

() -capture and group

\B -Matches where the specified characters are present,but not at the begninng Or End of the word r"\Bain" r"ain\B"

         2020-05-23

         2020-07-11

         2020-08-11

         so we  use "\d\d\d\d .\d\d.\d\d" ---"." character is used to find any character other than newline character.

```python
matches=re.finditer(r"\d\d\d\d.\d\d.\d\d",dates)

for match in matches:
  print(match)
```

1. We are looking for the  date in this format:

         2020-05-23

         2020-06-11

```python
matches=re.finditer(r"\d\d\d\d-\d\d-\d\d",dates)

for match in matches:
  print(match)
```
