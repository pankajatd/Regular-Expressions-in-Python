```python
def get_pattern_match(pattern,text):
  matches=re.findall(pattern,text)
  if matches:
    return matches[0]
get_pattern_match(r"\(age.*\n(.*)",text)
```

Writing a Function to get allthe above Information in a Single Function

```python
def get_personal_information(text):
  age=get_pattern_match(r"\(age (\d+)\)",text)
  full_name=get_pattern_match(r"Born(.*)",text)
  birth_date=get_pattern_match(r"Born.*\n(.*)",text)
  birth_place=get_pattern_match(r"\(age.*\n(.*)",text)

  return {
          'age':int(age),
           'name':full_name.strip(),
           'birth_date':birth_date.strip(),
           'birth_place':birth_place.strip()
  }
get_personal_information(text)
```
