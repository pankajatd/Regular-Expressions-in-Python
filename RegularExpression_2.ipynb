**Regular Expression**

Identifying the 10-Digit phone number from the input string

```python
import re
```

```python
tesla_note =''' Elon musk's phone number  is 9991116666,call him if you have any questions on dodgecoin.Tesla's revenue is 40 billion '''
```

```python
matches=re.finditer(r"\d{10}",tesla_note)
for match in matches:
  print(match)
```

**Extract FY2021 and fy2020 from  the below Text**

```python
import re
```

```python
text =""" The gross cost of Operating lease vehicles in FY2021 Q1 was $4.85 billion.In previous quarter i.e is fy2020 Q4 it was $3 billion. """
```

```python
matches =re.finditer(r'FY(\d{4} Q[1-4])',text,flags=re.IGNORECASE)
for match in matches:
  print(match.group(0))
```

**Extract  FY2021 Q1 $4.85  and fy2020 Q4 $3 from  the below Text**

```python
import re
```

```python
text =""" The gross cost of Operating lease vehicles in FY2021 Q1 was $4.85 billion.In previous quarter i.e is fy2020 Q4 it was $3 billion."""
```

```python
matches =re.finditer(r"FY(\d{4} Q[1-4])[^\$]+\$([0-9\.]+)",text,flags=re.IGNORECASE)
for match in matches:
  print(match.group(2))
```
